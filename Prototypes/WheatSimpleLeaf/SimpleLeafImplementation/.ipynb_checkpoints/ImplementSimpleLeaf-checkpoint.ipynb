```python
def AddOrModifyCultivars(NewCultivarParams):
    ## Go through and overwrite the parameter values for existing cultivars
    ## or if cultivar is currently listed as an alias remove that. 
    ModifiedCultivarList = []
    CultivarsToModifyOrAdd = NewCultivarParams.index
    ParametersToModifyOrAdd = NewCultivarParams.columns
    CultivarGroup = findModel(Wheat,['Wheat','Cultivars'])
    Cultivars = findCultivars(CultivarGroup)
    for c in Cultivars:
        cult = findCultivar(CultivarGroup,c)
        try:
            Aliai = cult['Children']
            RetainedAliai = []
            for a in Aliai:
                if (a['Name'] not in CultivarsToModifyOrAdd) | (a["$type"] != "Models.Core.Alias, Models"):
                    RetainedAliai.append(a)
        except:
            Aliai = 'None'
        cult['Children'] = RetainedAliai
        cultName = cult['Name']
        #Then go through each cultivar and Add or Modity parameter commands
        ModifiedParameterList = []
        if cultName in CultivarsToModifyOrAdd:
            cultCommands = cult['Command']
            #Make a new param list to replace existing
            NewCommands = []
            #Overwrite commands we want to modify
            for com in cultCommands:
                paramName = com.split('=')[0].replace(' ','')
                if (paramName in ParametersToModifyOrAdd) and (~np.isnan(NewCultivarParams.loc[cultName,paramName])) :
                    newCommand = paramName + ' = ' + str(NewCultivarParams.loc[cultName,paramName])
                    NewCommands.append(newCommand)    
                    ModifiedParameterList.append(paramName)
                #Add params that are not in our modify table back into cultivar params as they were
                if paramName not in ParametersToModifyOrAdd: 
                    NewCommands.append(com)
            #Add commands that are not currently in cultivar
            for paramName in ParametersToModifyOrAdd:
                if (paramName not in ModifiedParameterList) and (~np.isnan(NewCultivarParams.loc[cultName,paramName])):
                    newCommand = paramName + ' = ' + str(NewCultivarParams.loc[cultName,paramName])
                    NewCommands.append(newCommand)    
            # Replace existing cultivar description with modified one
            cult['Command'] = NewCommands
            ModifiedCultivarList.append(cultName)
    
    # Add new cultivars into Just Added folder
    CultivarsToAdd = findModel(Wheat,['Wheat','Cultivars',"Just Added"])
    AddedCultivarList = []
    ## Go through and add in cultivars that do not currently exist
    for cultName in CultivarsToModifyOrAdd:
        if cultName not in ModifiedCultivarList:
            comToAdd = {"$type": "Models.PMF.Cultivar, Models",
                      "Command": [],
                      "Name": cultName,
                      }
            for paramName in ParametersToModifyOrAdd:
                if(~np.isnan(NewCultivarParams.loc[cultName,paramName])):
                    pCommand = paramName + ' = ' + str(NewCultivarParams.loc[cultName,paramName])
                    comToAdd['Command'].append(pCommand)
            CultivarsToAdd['Children'].append(comToAdd)
            AddedCultivarList.append(cultName)
            print(cultName + ' cultivar added to Just Added folder')
#     ## Add empty folder to put new cultivars into
#     CultivarGroup['Children'].append(CultivarsToAdd)
#     print('AfterAdds')
#     print(findCultivars(CultivarGroup))
    
    
# def AddCultivarsIfMissing(NewCultivarParams):
#     ## Go through and Add fitted parameters for those that are currently in as alisases
#     ExistingCultivarList = []
#     CultivarsAdd = NewCultivarParams.index
#     ParametersAdd = NewCultivarParams.columns
#     CultivarGroup = findModel(Wheat,['Wheat','Cultivars'])
#     Cultivars = findCultivars(CultivarGroup)
#     for c in Cultivars:
#         cult = findCultivar(CultivarGroup,c)
#         try:
#             Aliai = cult['Children']
#             RetainedAliai = []
#             for a in Aliai:
#                 if (a['Name'] not in CultivarsAdd) | (a["$type"] != "Models.Core.Alias, Models"):
#                     RetainedAliai.append(a)
#         except:
#             Aliai = 'None'
#         cult['Children'] = RetainedAliai
#         cultName = cult['Name']
#         #Then go through each cultivar and Add to list
#         if cultName in CultivarsAdd:
#             ExistingCultivarList.append(cultName)

#     # Create empty folder to put new cultivars into
#     try:
#         CultivarsToAdd = findModel(Wheat,['Wheat','Cultivars',"Just Added"])
#     except:
#         print('adding Just Added cultivar folder')
#         CultivarsToAdd = {"$type": "Models.PMF.CultivarFolder, Models",
#                       "Name": "Just Added",
#                       "Children": []}
#     AddedCultivarList = []
#     ## Go through and add in cultivars that do not currently exist
#     for cultName in CultivarsAdd:
#         if cultName not in ExistingCultivarList:
#             comToAdd = {"$type": "Models.PMF.Cultivar, Models",
#                       "Command": [],
#                       "Name": cultName,
#                       }
#             for paramName in ParametersAdd:
#                 pCommand = paramName + ' = ' + str(NewCultivarParams.loc[cultName,paramName])
#                 comToAdd['Command'].append(pCommand)
#             CultivarsToAdd['Children'].append(comToAdd)
#             AddedCultivarList.append(cultName)
#     ## Add empty folder to put new cultivars into
#     CultivarGroup['Children'].append(CultivarsToAdd)

def AddMaxLAR(CultivarMaxLAR):
    ## Go through add MaxLAR to each cultivar and remove phyllochron
    CultivarGroup = findModel(Wheat,['Wheat','Cultivars'])
    Cultivars = findCultivars(CultivarGroup)
    for cult in Cultivars:
        cultCommands = findCultivar(CultivarGroup,cult)['Command']
        pointer = '[Phenology].MaxLAR.FixedValue = '
        try:
            MaxLARcommand = pointer + str(CultivarMaxLAR.loc[cult,'MaxLAR'])
        except:
            MaxLARcommand = pointer + str(0.015) 
        cultCommands.append(MaxLARcommand)
        pointer = '[Phenology].PTQhf.FixedValue = '
        try:
            PTQhfcommand = pointer + str(CultivarMaxLAR.loc[cult,'PTQhf'])
        except:
            PTQhfcommand = pointer + str(0.9)
        cultCommands.append(PTQhfcommand)
        pos=0
        for com in cultCommands:
            if '[Phenology].Phyllochron.BasePhyllochron.FixedValue' in com:
                del cultCommands[pos]
            pos+=1

def findNextChild(Parent,ChildName):
    if len(Parent['Children']) >0:
        for child in range(len(Parent['Children'])):
            if Parent['Children'][child]['Name'] == ChildName:
                return Parent['Children'][child]
    else:
        return Parent[ChildName]

def findModel(Parent,PathElements):
    for pe in PathElements:
        Parent = findNextChild(Parent,pe)
    return Parent

def replaceModel(Parent,modelPath,New):
    PathElements = modelPath.split('.')
    try:
        test = findModel(Parent,PathElements[:-1])[PathElements[-1]]
        findModel(Parent,PathElements[:-1])[PathElements[-1]] = New
    except:
        try:
            pos = 0
            for kid in findModel(Parent,PathElements[:-1])['Children']:
                if kid['Name'] == PathElements[-1]:
                    findModel(Parent,PathElements[:-1])['Children'][pos] = New
                    break
                pos +=1
        except:   
            print('Could not find parent node of model to over write for ' + modelPath)
            raise
        
def removeModel(Parent,modelPath):
    PathElements = modelPath.split('.')
    Parent = findModel(Parent,PathElements[:-1])
    pos = 0
    found = False
    for c in Parent['Children']:
        if c['Name'] == PathElements[-1]:
            del Parent['Children'][pos]
            found = True
            break
        pos += 1
    if found == False:
        print('Failed to find ' + PathElements[-1] + ' to delete')
    # Remove cultivar overwrites
    CultivarGroup = findModel(Wheat,['Wheat','Cultivars'])
    Cultivars = findCultivars(CultivarGroup)
    for c in Cultivars:
        cult = findCultivar(CultivarGroup,c)
        try:
            cultCommands = cult['Command']
            #Overwrite commands we want to modify
            for com in cultCommands:
                paramName = com.split('=')[0].replace(' ','')
                if paramName == modelPath:
                    del(com)
        except:
            everyting = 'good'
    
            
def addModel(Parent,modelPath,New):
    PathElements = modelPath.split('.')
    Parent = findModel(Parent,PathElements)
    if Parent == None:
        print('Could not find parent model ' + modelPath + ' to Add new model to.  Dont include the name of the new models name in the path')
    if isinstance(New,dict):
        NewDict = New
    else:
        NewDict = json.loads(New)
    Parent['Children'].append(NewDict)

def renameModel(Parent,modelPath,NewName):
    PathElements = modelPath.split('.')
    Parent = findModel(Parent,PathElements)
    Parent['Name'] = NewName
    
def ImplementNewModel(NewParams):
    for n in NewParams.index:
        if NewParams.loc[n,'Action'] == 'Remove':
            print(NewParams.loc[n,'Action']+'ing',NewParams.loc[n,'Path'])
            removeModel(Wheat,NewParams.loc[n,'Path'])
        elif NewParams.loc[n,'Action'] == 'Overwrite':
            print(NewParams.loc[n,'Action']+'ing',NewParams.loc[n,'Path'])
            try:
                with open(NewParams.loc[n,'NewValue'],'r') as newValueJSON:
                    newValue = json.load(newValueJSON)
                    newValueJSON.close()
            except:
                if isinstance(NewParams.loc[n,'NewValue'],numbers.Number):
                    newValue = NewParams.loc[n,'NewValue']
                else:
                    newValue = json.loads(NewParams.loc[n,'NewValue'])
            replaceModel(Wheat,
                         NewParams.loc[n,'Path'],
                         newValue)
        elif NewParams.loc[n,'Action'] == 'Add':
            print(NewParams.loc[n,'Action']+'ing',NewParams.loc[n,'Path'])
            try:
                with open(NewParams.loc[n,'NewValue'],'r') as newValueJSON:
                    newValue = json.load(newValueJSON)
                    newValueJSON.close()
            except:
                newValue = NewParams.loc[n,'NewValue']
            addModel(Wheat,NewParams.loc[n,'Path'], newValue)
        elif NewParams.loc[n,'Action'] == 'Rename':
            print(NewParams.loc[n,'Action']+'ing',NewParams.loc[n,'Path'])
            renameModel(Wheat,NewParams.loc[n,'Path'], NewParams.loc[n,'NewValue'])

        else:
            print('Model changes failed')

def findCultivar(CultivarGroup,CultivarName):
    for folder in CultivarGroup['Children']:
        for cultivar in folder['Children']:
            if cultivar['Name'] == CultivarName:
                return cultivar

def findCultivars(CG):
    CultivarList = []
    for folder in CG['Children']:
        for cultivar in folder['Children']:
            CultivarList.append(cultivar['Name'])
    return CultivarList

def findCultivarsAndAlisas(CG):
    CultivarList = []
    for folder in CG['Children']:
        for cultivar in folder['Children']:
            CultivarList.append(cultivar['Name'])
            for alias in cultivar['Children']:
                CultivarList.append(alias['Name'])
    return CultivarList

def AddFLNParams():
    ## Go through LN Parmeters for existing cultivars and convert to FLN params
    mln = '[Phenology].MinimumLeafNumber.FixedValue'
    vs = '[Phenology].VrnSensitivity.FixedValue'
    ps = '[Phenology].PpSensitivity.FixedValue'
    CultivarGroup = findModel(Wheat,['Wheat','Cultivars'])
    Cultivars = findCultivars(CultivarGroup)
    for cu in Cultivars:
        Cultivar = findCultivar(CultivarGroup,cu)
        LV = '0';
        pos = 0
        for co in Cultivar['Command']:
            if co[:len(mln)] == mln:
                LV = co.replace(' ', '').split('=')[1]
                new = '[Phenology].CAMP.FLNparams.MinLN = ' + LV
                Cultivar['Command'].append(new)
                del Cultivar['Command'][pos]
                break
            pos +=1
        if LV == '0':
            LV = '7' #use default if no value specified
            new = '[Phenology].CAMP.FLNparams.MinLN = ' + LV
            Cultivar['Command'].append(new)
        
        LN = '0'
        pos = 0
        for co in Cultivar['Command']:
            if co[:len(ps)] == ps:
                value = co.replace(' ', '').split('=')[1]
                PpLN = str(float(value))
                new = '[Phenology].CAMP.FLNparams.PpLN = ' + PpLN
                Cultivar['Command'].append(new)
                del Cultivar['Command'][pos]
                break
            pos +=1 
        # if PpLN == '0':
        #     PpLN = 0  # 0 is default for vern sensitivity
        #     new = '[Phenology].CAMP.FLNparams.LN = ' + LN 
        #     Cultivar['Command'].append(new)
        
        pos = 0
        for co in Cultivar['Command']:
            if co[:len(vs)] == vs:
                VernLN = co.replace(' ', '').split('=')[1]
                new = '[Phenology].CAMP.FLNparams.VrnLN = ' + VernLN
                Cultivar['Command'].append(new)
                del Cultivar['Command'][pos]
                break
            pos +=1 
        # if SV == '0':
        #     SV = str(float(LV)  + 3) # 3 is default for Pp sensitivity
        #     new = '[Phenology].CAMP.FLNparams.SV = ' + SV 
        #     Cultivar['Command'].append(new)
        
        SN = str(0.0)
        new = '[Phenology].CAMP.FLNparams.VxPLN = ' + SN
        Cultivar['Command'].append(new)    
        
def SetFLNOfAllTo7():
    ## Go through and overwrite the parameter values so all cultivars always have a FLN of 7
    FLNParamNames = ['[Phenology].MinimumLeafNumber.FixedValue',
             '[Phenology].PpSensitivity.FixedValue',
             '[Phenology].VrnSensitivity.FixedValue']
    FLNParams = pd.DataFrame(index = FLNParamNames,data = [7.0,0.0,0.0],columns=['Value'])
    ModifiedCultivarList = []
    CultivarGroup = findModel(Wheat,['Wheat','Cultivars'])
    Cultivars = findCultivars(CultivarGroup)
    for c in Cultivars:
        cult = findCultivar(CultivarGroup,c)
        #Go through each cultivar and Modity FLN Parameters
        ParametersToModifyOrAdd = FLNParams.index
        ModifiedParameterList = []
        cultCommands = cult['Command']
        #Make a new param list to replace existing
        NewCommands = []
        #Overwrite commands we want to modify
        for com in cultCommands:
            paramName = com.split('=')[0].replace(' ','')
            if paramName in ParametersToModifyOrAdd:
                newCommand = paramName + ' = ' + str(FLNParams.loc[paramName,'Value'])
                NewCommands.append(newCommand)    
                ModifiedParameterList.append(paramName)
            #Add params that are not in our modify table back into cultivar params as they were
            if paramName not in ParametersToModifyOrAdd: 
                NewCommands.append(com)
        #Add commands that are not currently in cultivar
        for paramName in ParametersToModifyOrAdd:
            if paramName not in ModifiedParameterList:
                newCommand = paramName + ' = ' + str(FLNParams.loc[paramName,'Value'])
                NewCommands.append(newCommand)    
        # Replace existing cultivar description with modified one
        cult['Command'] = NewCommands
        
def WipeCultivarOverwrites():
    CultivarGroup = findModel(Wheat,['Wheat','Cultivars'])
    Cultivars = findCultivars(CultivarGroup)
    for c in Cultivars:
        cult = findCultivar(CultivarGroup,c)
        cult['Command'] = []
```

```python
with open('C:\GitHubRepos\ApsimX\Models\Resources\Wheat.json','r') as WheatJSON:
    Wheat = json.load(WheatJSON)
    WheatJSON.close()
    
toRemove = ['Advantage', 'Alberic', 'Anlace', 'Annuello', 'Arrino', 'Arrivato',
       'Arrow', 'Aspiring', 'Aurora', 'Babbler', 'Barham', 'Baxter',
       'Bellaroi', 'BonnieRock', 'Bowerbird', 'Bowie', 'Brennan',
       'Buckley', 'Burunga', 'Camm', 'Caparoi', 'Carinya', 'Carnamah',
       'Centaur', 'Chara', 'Chief', 'Cobra', 'Corack', 'Correll',
       'Dagger', 'Dart', 'Declic', 'Diamondbird', 'Dollarbird',
       'Einstein', 'Elmore', 'EmuRock', 'Endure', 'Envoy', 'Espada',
       'Estoc', 'Exceed', 'Frame', 'Gazelle', 'Giles', 'Goroke',
       'Guardian', 'Halberd', 'Harper', 'Hunter', 'Hyperno',
       'Impala', 'Impose', 'Justica', 'Kalka', 'Katana', 'Kord',
       'Krichauff', 'Kukri', 'Lorikeet', 'Majestic', 'Merlin', 'Meteor',
       'Mitre', 'Option', 'Pennant', 'Phantom', 'Preston',
       'Pugsley', 'Regency', 'Richmond', 'Robigus', 'Rubric',
       'Sage', 'Saintly', 'Saphire', 'Savannah', 'Scenario', 'Sentinel',
       'Shield', 'Silverstar', 'Solstice', 'Stiletto',
       'Sunguard', 'Sunlin', 'Sunvale', 'Tamarinrock', 'Tamaroi',
       'Tanker', 'Tribute', 'Waagin', 'Wallup', 'Weston',
       'Westonia', 'Wollaroi', 'Wylah', 'Yallaroi', 'Yawa', 'Zen',
       'Zippy', 'Adagio']
    
CultivarGroup = findModel(Wheat,['Wheat','Cultivars'])

def removeAliases(toRemove,CG):
    AliaiRemoved = []
    CultivarsRemoved = []
    CultivarsNotRemoved = []
    for folder in CG['Children']:
        cpos = 0
        for cultivar in folder['Children']:
            try:
                toDel = []
                apos=0
                for alias in cultivar['Children']:
                    print(alias['Name'])
                    if alias['Name'] in toRemove:
                        AliaiRemoved.append(alias['Name'])
                        toDel.append(apos)
                    apos +=1
                if len(toDel)>0:
                    for apos in reversed(toDel):
                        del cultivar['Children'][apos]
                if cultivar['Name'] in toRemove:
                    if apos == 0:
                        del folder['Children'][cpos]
                        CultivarsRemoved.append(cultivar['Name'])
                    else:
                        CultivarsNotRemoved.append(cultivar['Name'])
            except:
                print(cultivar['Name'] + " has not children property")
            cpos +=1
    AliaiRemoved.sort()
    CultivarsRemoved.sort()
    CultivarsNotRemoved.sort()
    
    return AliaiRemoved, CultivarsRemoved, CultivarsNotRemoved

AliaiRemoved, CultivarsRemoved, CultivarsNotRemoved = removeAliases(toRemove,CultivarGroup)

# ## Save the altered wheat.json file
with open('C:\GitHubRepos\ApsimX\Models\Resources\Wheat.json','w') as WheatJSON:
    json.dump(Wheat,WheatJSON,indent=2)
```
